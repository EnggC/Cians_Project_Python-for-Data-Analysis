Understanding .pct_change vs .pct_change()
When you write:

python
returns[tick + ' returns'] = bank_stocks[tick]['Close'].pct_change
you’re assigning the method object itself to your returns DataFrame column, not its output. To actually compute percentage changes, you need to call the method with parentheses:

python
returns[tick + ' returns'] = bank_stocks[tick]['Close'].pct_change()
What’s really happening
.pct_change

This is a reference to the function.

It won’t perform any calculation.

If you inspect returns afterward, the column contains a pointer to the method, not numbers.

.pct_change()

This executes the function with its default argument (periods=1).

Assigning this gives you the actual return values.

Why You’re Getting “TypeError: 'module' object is not callable”
That error means that when you write

python
plt.legend()
Python thinks plt.legend is a module, not the legend() function you expect. Trying to “call” a module produces exactly that TypeError.

Common Causes & Fixes
Wrong import alias If you did

python
import matplotlib as plt
then plt is the top-level matplotlib module, and plt.legend points to the matplotlib.legend submodule.

Fix

python
import matplotlib.pyplot as plt
ensures plt.legend refers to the plotting function.
