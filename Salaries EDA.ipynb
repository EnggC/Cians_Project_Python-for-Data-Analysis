```python
# Convert 'BasePay' to numeric, forcing errors to NaN
sal['BasePay'] = pd.to_numeric(sal['BasePay'], errors='coerce')

# Now group by 'Year' and calculate the mean of 'BasePay'
mean_basepay_by_year = sal.groupby('Year')['BasePay'].mean()
print(mean_basepay_by_year)
```
